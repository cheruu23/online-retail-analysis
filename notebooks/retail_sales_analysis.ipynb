```python
# Check the number of duplicate rows
num_duplicates = retail_df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")

# Remove duplicates and update the dataframe in place
retail_df.drop_duplicates(inplace=True)

# Verify new shape
print(f"New shape after dropping duplicates: {retail_df.shape}")
```

```python
# View rows where Description is null
missing_desc = retail_df[retail_df['Description'].isnull()]
print(f"Total rows with missing description: {len(missing_desc)}")
missing_desc.head()
```

```python
# Investigate rows with missing CustomerID
missing_cust = retail_df[retail_df['CustomerID'].isnull()]
print(f"Shape of records missing CustomerID: {missing_cust.shape}")

# Check if these transactions have valid revenue or quantities
missing_cust.describe()
```

```python
# Convert InvoiceNo to string to safely check for 'C' prefix
retail_df['InvoiceNo'] = retail_df['InvoiceNo'].astype(str)

# Create a filtered dataframe excluding cancelled invoices for sales analysis
retail_clean_df = retail_df[~retail_df['InvoiceNo'].str.startswith('C')].copy()

print(f"Original dataset rows: {len(retail_df)}")
print(f"Cleaned dataset rows (excluding cancellations): {len(retail_clean_df)}")
```

```python
# Filter for negative quantities
returns_df = retail_clean_df[retail_clean_df['Quantity'] < 0]
print(f"Number of return transactions: {len(returns_df)}")
returns_df.head()
```

```python
# Check for zero or negative prices
invalid_prices = retail_clean_df[retail_clean_df['UnitPrice'] <= 0]
print(f"Number of rows with zero or negative UnitPrice: {len(invalid_prices)}")
invalid_prices
```

```python
retail_clean_df = retail_clean_df[retail_clean_df['UnitPrice'] > 0]
```

```python
# Final validation check
print("Final Clean Dataset Shape:", retail_clean_df.shape)
print("\nRemaining Missing Values:")
print(retail_clean_df.isnull().sum())
print("\nSummary Statistics:")
retail_clean_df.describe()
```

## visualization using matplot 

```python
# Get value counts of countries
retail_df_country = retail_df['Country'].value_counts()

# Create the plot for the top 10 countries
fig, ax = plt.subplots(figsize=(10, 6))
retail_df_country.head(10).plot(kind='bar', color='skyblue', ax=ax)

# Add titles and labels
ax.set_title("Top 10 Countries by Transaction Count", fontsize=14)
ax.set_xlabel("Country", fontsize=12)
ax.set_ylabel("Number of Transactions", fontsize=12)

# Rotate country names so they don't overlap
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
